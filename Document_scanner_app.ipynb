## Document Scanner Pro (AI-Powered Web App)

Document Scanner Pro is a browser-based application that converts any document image into a clean, scanned version using advanced computer vision techniques powered by OpenCV.js.

## Features
Automatic Document Detection
Detects document edges using Canny edge detection and contour analysis.
Perspective Correction
Applies a four-point transform to fix skewed or tilted images.
Background Removal
Removes dark or noisy backgrounds and converts them into a clean white surface.
Image Enhancement
Improves clarity using contrast adjustment, brightness control, CLAHE, and denoising.
Black & White Mode
Optional high-quality B&W output for better readability of text documents.
Rotation Support
Rotate scanned documents (90°, 180°) easily.
Multiple Export Options
Download scanned documents in:
PNG (high quality)
JPEG (compressed)
PDF (print-ready)
Real-Time Processing in Browser
Fully client-side processing — no data is uploaded to any server.
Interactive UI
Includes sliders for brightness/contrast, progress tracking, and preview panels.

## Technologies Used
HTML, CSS, JavaScript
OpenCV.js (Computer Vision)
Canvas API
jsPDF (for PDF export)

## How It Works
Upload an image of a document
The system detects edges and finds document boundaries
Perspective correction is applied
Background is cleaned and image is enhanced
Download the final scanned document

## Privacy Friendly
100% offline processing
No server upload — your data stays (secure) in your browser
